Crie um programa que leia algo pelo teclado e mostre na tela o seu tipo primitivo e todas as informações possiveis sobre ele.<br>

```python
valor = input("Digite qualquer coisa: ")
print(f"O tipo primitivo de {valor} é ", type(valor))
```

```python
valor = input("Digite qualquer coisa: ")
print(f"O tipo primitivo de {valor} é ", type(valor))
print(f"{valor} é um número? ", valor.isnumeric())
print(f"{valor} é alfanumérico? ", valor.isalnum())
print(f"{valor} contem espaços? ", valor.isspace())
print(f"{valor} está em minúsculas", valor.islower())
print(f"{valor} está em maiúsculas?", valor.isupper())
print(f"{valor} está capitalizado?", valor.istitle())
```
