# Exercício 5 - Antecessor e Sucessor
Crie um programa que leia um número inteiro e mostre na tela seu antecessor e sucessor.<br>
Curso em Vídeo | [Desafio e resolução](https://youtu.be/664e0G_S9nU?si=PA9p2mwIU9uCdPFJ)

```python
numero = int(input("Digite um número: ")) # Solicita um número ao usuário
antecessor = numero - 1 # Pega o valor de (numero) e diminui 1
sucessor = numero + 1 #Pega o valor de (numero) e soma 1
print(f"O antecessor de {numero} é {antecessor} e o sucessor é {sucessor}") 
```
