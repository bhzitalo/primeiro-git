# Exercício 6 - Dobro, Triplo e Raiz Quadrada
Crie um algoritmo que leia um número e mostre seu dobro, triplo e raiz quadrada.<br>
Curso em Vídeo | [Desafio e resolução](https://youtu.be/mqcNw_dhl8I?si=WbwX_4Xv6jMGV4Gd)

```python
numero = int(input("Digite um número: "))
print(f"O dobro de {numero} é {numero*2}")
print(f"O triplo de {numero} é {numero*3}")
print(f"A raiz quadrada de {numero} é {numero **(1/2):.2}")
```
