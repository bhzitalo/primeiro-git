Crie um programa que leia duas notas de um aluno, calcule e mostre sua média.<br>

```python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
print(f"A média do aluno é {(nota1 + nota2) / 2}")
```
