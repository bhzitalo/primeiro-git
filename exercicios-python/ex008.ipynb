# Exercício 8 - Conversor de Medidas
Crie um programa que leia um valor em metros e exiba convertido em centímetros e milímetros.<br>
Curso em Vídeo | [Desafio e Resolução](https://youtu.be/KjcdG05EAZc?si=uIX4kpx3FdJBgZN_)

```python
numero = float(input("Digite um número em metros: "))
print(f"O valor de {numero} metros equivale a {numero*100:.0f} centimetros")
print(f"O valor de {numero} metros equivale a {numero*1000:.0f} milímetros")
```

```python
numero = float(input("Digite um número em metros: "))
cm = numero * 100
mm = numero * 1000
print(f"{numero} metros equivale a {cm:.0f} cm")
print(f"{numero} metros equivale a {mm:.0f} mm")
```
