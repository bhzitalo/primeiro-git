# Exercício 9 | Tabuada
Crie um programa que leia um número inteiro e mostre na tela a sua tabuada.<br>
Curso em Vídeo | [Desafio e resolução](https://youtu.be/qajq3SI0QQs?si=aTwNpkUWzUS0QXp7)

```python
numero = int(input("Digite um número:"))
print(f"{numero} x 1 = {numero*1}")
print(f"{numero} x 2 = {numero*2}")
print(f"{numero} x 3 = {numero*3}")
print(f"{numero} x 4 = {numero*4}")
print(f"{numero} x 5 = {numero*5}")
print(f"{numero} x 6 = {numero*6}")
print(f"{numero} x 7 = {numero*7}")
print(f"{numero} x 8 = {numero*8}")
print(f"{numero} x 9 = {numero*9}")
print(f"{numero} x 10 = {numero*10}")
```

```python
numero = int(input("Digite um número: "))
contador = 1

while contador <= 10: # enquanto o contador for menor ou igual a 10, faça:
    multiplicador = numero * contador
    print(f"{numero} x {contador} = {multiplicador}")
    contador = contador + 1 # contador recebe +1 até chegar a 10
```

```python
numero = int(input("Digite um número: "))
for contador in range(1,11):
    print(f"{numero} x {contador} = {numero * contador}")
```
