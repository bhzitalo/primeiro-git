# Exercício Python 10 - Conversor de Moedas
Crie um programa que leia quanto dinheiro uma pessoa tem na carteira e mostre quantos dólares ela pode comprar.<br>
Curso em Vídeo | [Desafio e resolução](https://youtu.be/xM4AX3Lp2mo?si=d3LloqB41mReQSKL)

```python
dinheiro = float(input("Quanto dinheiro você tem na carteira?"))
cotacao = 5.16

print(f"Com R$ {dinheiro:.2f} você pode comprar US$ {dinheiro/cotacao:.2f}")
```

```python
dinheiro = float(input("Quanto dinheiro você tem na carteira?"))
cotacao = float(input("Qual a cotação do dólar?"))

print(f"Dinheiro na carteira: R$ {dinheiro:.2f}") 
print(f"Cotação do dólar: R$ {cotacao:.2f}") 
print(f"Com R$ {dinheiro:.2f} você pode comprar US$ {dinheiro/cotacao:.2f}")
```
