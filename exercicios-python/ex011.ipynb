# Exercício Python #11 - Pintando Parede
Faça um programa que leia a largura e a altura de uma parede em metros, calcule a sua área e a quantidade de tinta necessária para pintá-la, sabendo que cada litro de tinta pinta uma área de 2 metros quadrados.<br>
Curso em Vídeo | [Desafio e resolução](https://youtu.be/mzSJpn9ldt4?si=uB2BHQt5PqkWC4iw)




```python
largura = float(input("Digite a largura da parede em metros: "))
altura = float(input("Digite a altura da parede em metros: "))
areaParede = largura * altura
tintaParede = areaParede / 2

print(f"Largura: {largura} m")
print(f"Altura: {altura} m")
print(f"Área: {areaParede:.2f} m²")
print(f"Tinta necessária: {tintaParede:.2f} litros")
```
