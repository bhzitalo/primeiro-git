# Exercício Python #12 - Calculando Descontos
Exercício Python 012: Faça um algoritmo que leia o preço de um produto e mostre seu novo preço, com 5% de desconto.<br>
Curso em Vídeo | [Desafio e resolução](https://youtu.be/4MAmKOT9FeU?si=Is2IEN41rksPJYQv)

```python
precoProduto = float(input("Digite o preço do produto: R$"))
desconto = precoProduto * 5 / 100
precoFinal = precoProduto - desconto


print(f"Preço original: R$ {precoProduto:.2f}")
print(f"Desconto: R$ {desconto:.2f}")
print(f"Preço com desconto: R$ {precoFinal:.2f}")
```

```python
precoProduto = float(input("Digite o proço do produto: R$"))
percentutalDesconto = float(input("Digite o percentual de desconto: "))
desconto = precoProduto * percentutalDesconto / 100
precoFinal = precoProduto - desconto

print(f"Preço original: R$ {precoProduto:.2f}")
print(f"Desconto: {percentutalDesconto}% = R$ {desconto:.2f}")
print(f"Preço com desconto: R$ {precoFinal:.2f}")
```
